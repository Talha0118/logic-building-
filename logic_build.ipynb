```python
for row in range(0,5):
    for col in range(0,5):
        print("*",end=' ')
    print()
```

```python
for row in range(0,5):
    for col in range(0,row+1):
        print('*',end=' ')
        
    print()
```

```python
for row in range(0,5):
    for col in range(row,5):
        print('*',end=' ')
        
    print()
```

```python
for row in range(0,5):
    for col in range(0,row+1):
        print(col+1,end=' ')
        
    print()
```

```python
for row in range(0,5):
    for col in range(row,5):
        print(col+1,end=' ')
        
    print()
```

```python

for row in range(0,5):
    ch = 65
    for col in range(0,row+1):
        print(chr(ch+col),end=' ')
        
    print()
```
